# 5 Real World Product Uses of Deep Learning.


1. Youtube:

- Uses Deep Learning in Video Recommendation System.
- Two Stage Architecture: First uses the Neural Networks for recommendation system, then rank them on basis of user satisfaction, watch time , etc.
- Process Video Titles, descriptions using NLP.
- Uses CNN and Computer Vision to analyse video frames and thumbnail and visual content to understand context and categories it accurately.


2. Zomato

- Uses LSTM for Food Preperation Time Prediction by processing complex time series data related to restaurant footfall.
- During the delivery process, Zomato employs Convolutional Neural Networks (CNNs) for automated safety audits.

3. Tesla

- Uses DL for FULL Self Driving capabilities.
- Uses HydraNet Architecture , enabling 50 NNs to process tasks like lane detection, traffic light recognition and 3D object tracking.
- CNN and YOLO are applied in Manufacturing for automated quailty inspection. 

4. Robots

- CNNs enable computer Vision to detect objects and recognize face using camers and LiDAR sensors.
- RNNs and transformers facilitates NLP, enabling robots to understand speech, parse commands.

5. WhatsApp

- Uses DL to combat with misinformation and fake News.
- Hybrid model of BERT, LSTM and Deep Neural Networks helps to analyze message content to classify it as real or fake.
- RNN generates text responses to propose instant, pre-written replies , saving users time.
