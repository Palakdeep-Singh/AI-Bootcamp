# Q1: What is AI ?

AI means machine performing the tasks that require human intellegence. AI involves learning, reasoning, decision-making.

# Q2: What is ML ?

Machine Learning is a branch of AI where algorithms learn patterns from data instead of being explicitly programmed, allowing them to make predictions, classifications, recommendations, or other decisions on unseen data.

# O3: What is DL ?

Deep Learning is a subset of Machine Learning that uses deep neural networks (neural networks with multiple layers) to automatically learn complex patterns from large amounts of data.

# Q4: Where is DL used ?

DL is used in Computer Vision, Natural Language Processing, Speech Recognition, Recommendation Systems, Healthcare, Autonomous Vehicles, Robotics.

# Q5: Why is DL successful now ?

DL is successful now because we have alot of data in today's world. Neural networks are fed with a bunch of data for proper functioning, and also require good hardwares for computation power which is available now a days. Better algorithms and software frameworks.
